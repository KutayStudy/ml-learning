#### 56. Generate a generic 2D Gaussian-like array (★★☆)

````python
x = np.linspace(-1, 1, 10)
y = np.linspace(-1, 1, 10)
co, mu = 1.0, 0.0
gx = np.exp(-((x - mu) ** 2 / (2.0 * co**2)))
gy = np.exp(-((y - mu) ** 2 / (2.0 * co**2)))
G = gy[:, None] * gx[None, :]
print(G)
````
