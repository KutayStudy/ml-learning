# S6 — Broadcasting, Axis and NumPy Linear Algebra

````python
import numpy as np  # Import NumPy
````

## Broadcasting

````python
A = np.array([[10], [20], [30]])  # Shape: (3, 1)
B = np.array([[1, 2, 3, 4]])        # Shape: (1, 4)

result = A + B  # Broadcast to shape (3, 4)

print(A.shape, B.shape, result.shape)
print(result)
````

````python
A = np.zeros((5, 3))
row_values = np.array([10, 20, 30])
column_values = np.array([10, 20, 30, 40, 50])[:, None]

print(A + row_values)     # Apply values to each row
print(A + column_values)  # Apply one value to each row
````

````python
x = np.arange(4)

print(x.shape)
print(x[:, None].shape)
print(x[None, :].shape)
print(x[:, None] + x[None, :])  # Outer sum
````

````python
A = np.arange(8).reshape(2, 1, 4)
B = np.array([[10], [20], [30]])

result = A + B  # Interpret B as shape (1, 3, 1)

print(A.shape, B.shape, result.shape)
print(result)
````

## Axis and keepdims

````python
X = np.array([
    [1, 2, 3],
    [10, 20, 30]
])

print(X.mean())
print(X.mean(axis=0))
print(X.mean(axis=1))
print(X.mean(axis=-1))
````

````python
row_means = X.mean(axis=1, keepdims=True)
row_std = X.std(axis=1, keepdims=True)
standardized = (X - row_means) / row_std

print(row_means)
print(row_means.shape)
print(X - row_means)
print(standardized)
print(standardized.mean(axis=1))
print(standardized.std(axis=1))
````

````python
A = np.zeros((2, 3, 4))

print(A.mean(axis=0).shape)
print(A.mean(axis=1).shape)
print(A.mean(axis=2).shape)
print(A.mean(axis=-1).shape)
print(A.mean(axis=1, keepdims=True).shape)
````

````python
A = np.arange(24).reshape(2, 3, 4)

print(A.sum(axis=(1, 2)))
print(A.sum(axis=(1, 2)).shape)
print(A.sum(axis=(1, 2), keepdims=True))
print(A.sum(axis=(1, 2), keepdims=True).shape)
````

````python
A = np.array([
    [5, 2, 8],
    [1, 9, 4]
])

print(A.max(axis=0))
print(A.max(axis=1))
print(A.argmin(axis=0))
print(A.argmax(axis=1))
print(A.argmax(axis=1, keepdims=True))
````

## np.linalg.norm

````python
v = np.array([3.0, 4.0])

print(np.linalg.norm(v))
print(v / np.linalg.norm(v))
````

````python
X = np.array([
    [3.0, 4.0],
    [5.0, 12.0]
])

row_norms = np.linalg.norm(X, axis=1, keepdims=True)
normalized = X / row_norms

print(row_norms)
print(normalized)
print(np.linalg.norm(normalized, axis=1))
````

````python
X = np.array([
    [0.0, 0.0],
    [3.0, 4.0]
])

norms = np.linalg.norm(X, axis=1, keepdims=True)
safe_norms = np.where(norms == 0, 1, norms)

print(X / safe_norms)
````

## solve and inv

````python
A = np.array([
    [2.0, 1.0],
    [1.0, 3.0]
])
b = np.array([5.0, 6.0])

x = np.linalg.solve(A, b)

print(x)
print(A @ x)
print(np.allclose(A @ x, b))
````

````python
A_inverse = np.linalg.inv(A)

print(A_inverse)
print(A @ A_inverse)
print(np.allclose(x, A_inverse @ b))
````

````python
B = np.array([
    [5.0, 1.0],
    [6.0, 0.0]
])

X = np.linalg.solve(A, B)

print(X)
print(np.allclose(A @ X, B))
````

## matrix_rank and det

````python
full_rank = np.array([
    [1.0, 2.0],
    [3.0, 4.0]
])

rank_deficient = np.array([
    [1.0, 2.0],
    [2.0, 4.0]
])

print(np.linalg.matrix_rank(full_rank))
print(np.linalg.matrix_rank(rank_deficient))
print(np.linalg.det(full_rank))
print(np.linalg.det(rank_deficient))
````

````python
matrices = np.array([
    [[1.0, 0.0], [0.0, 1.0]],
    [[1.0, 2.0], [2.0, 4.0]]
])

print(np.linalg.matrix_rank(matrices))
print(np.linalg.det(matrices))
````

## eig and eigh

````python
A = np.array([
    [4.0, 1.0],
    [2.0, 3.0]
])

eigenvalues, V = np.linalg.eig(A)
Lambda = np.diag(eigenvalues)

print(eigenvalues)
print(V)
print(np.allclose(A @ V, V @ Lambda))
````

````python
A_reconstructed = V @ Lambda @ np.linalg.inv(V)

print(A_reconstructed)
print(np.allclose(A, A_reconstructed))
````

````python
S = np.array([
    [2.0, 1.0],
    [1.0, 3.0]
])

eigenvalues, Q = np.linalg.eigh(S)
Lambda = np.diag(eigenvalues)

print(eigenvalues)
print(Q.T @ Q)
print(np.allclose(S, Q @ Lambda @ Q.T))
````

## lstsq

````python
A = np.array([
    [1.0, 1.0],
    [1.0, 2.0],
    [1.0, 3.0]
])
b = np.array([1.0, 2.0, 2.0])

x, residuals, rank, singular_values = np.linalg.lstsq(A, b, rcond=None)

print(x)
print(A @ x)
print(b - A @ x)
print(rank)
print(singular_values)
````
