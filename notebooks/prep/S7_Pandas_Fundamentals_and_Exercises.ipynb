## 3. Boolean Indexing and `.query()`

```python
df[condition]
df.loc[condition, ["col1", "col2"]]
```

```python
(cond1) & (cond2)   # and
(cond1) | (cond2)   # or
~condition          # not
```

Useful patterns:

```python
df["column"].isin(["A", "B"])
df.query("salary >= 80000 and active")
df.query("salary >= @minimum_salary")
```

**Trick:** use `&` / `|`, not Python's `and` / `or`.

## 5. Reading CSV Files

```python
pd.read_csv(
    path_or_buffer,
    dtype={
        "employee_id": "string",
        "department": "category"
    },
    parse_dates=["start_date"]
)
```

**Tricks**

- IDs such as `001` should usually be strings.
- Parse dates while loading when possible.
- Check the result immediately with `.dtypes`.
- Reusing a `StringIO` object: call `csv_data.seek(0)` first.


````python
from io import StringIO

csv_data = StringIO("""
employee_id,name,department,salary,start_date,active
001,Ada,ML,72000,2025-09-15,True
002,Bora,Backend,85000,2024-03-10,True
003,Cem,ML,68000,2026-01-20,False
004,Deniz,Backend,92000,2023-11-05,True
005,Ece,Data,79000,2025-06-12,True
006,Fırat,Data,98000,2022-08-25,False
""".strip())
````

### Exercise 18 — ★☆☆
Read the CSV without `dtype` or `parse_dates`, then inspect both the table and inferred data types.


````python
csv_data.seek(0)
raw_employees = pd.read_csv(csv_data)
raw_employees
````

````python
raw_employees.dtypes
````

> Without an explicit `dtype`, `001` becomes `1`. Without `parse_dates`, `start_date` stays string-like.


### Exercise 19 — ★★☆
Read the same CSV using the following types: `employee_id → string`, `department → category`, `salary → int64`, `active → bool`, and parse `start_date` as a date.


````python
csv_data.seek(0)

employees_csv = pd.read_csv(
    csv_data,
    dtype={
        "employee_id": "string",
        "department": "category",
        "salary": "int64",
        "active": "bool"
    },
    parse_dates=["start_date"]
)

employees_csv.dtypes
````

Datetime columns support the `.dt` accessor:

```python

df["start_date"].dt.month
```


````python
employees_csv["start_date"].dt.year
````

````python
employees_csv["start_date"].dt.month
````

## 6. Grouping and Aggregation

Basic pattern:

```python
df.groupby("department")["salary"].mean()
```

Multiple functions:

```python
df.groupby("department").agg({
    "salary": ["mean", "std", "count"],
    "age": ["min", "max"]
})
```

Named aggregation:

```python
df.groupby("department", as_index=False).agg(
    average_salary=("salary", "mean"),
    employee_count=("name", "count")
)
```

**Tricks**

- `size()` counts rows.
- `count()` ignores missing values.
- Dictionary-based `.agg()` may create MultiIndex columns.
- Named aggregation gives clean output column names.


### Exercise 21 — ★☆☆
Calculate the average salary for each department.


````python
employees.groupby("department", as_index=True).agg({"salary": ["mean"]})
````

### Exercise 22 — ★☆☆

````python
employees.groupby("department", as_index=True).size()
````

### Exercise 23 — ★★☆
For each department, calculate `mean`, `std`, and `count` for the `salary` column using dictionary-based `.agg()`.


````python
employees.groupby("department", as_index=True).agg({"salary": ["mean", "std", "count"]})
````

### Exercise 24 — ★★☆
For each department, calculate `mean`, `min`, and `max` for both `salary` and `age`.


````python
employees.groupby("department", as_index=True).agg({"salary": ["mean", "min", "max"], "age": ["mean", "min", "max"]})
````

### Exercise 25 — ★★☆
Using named aggregation, create a result with `department`, `average_salary`, `highest_salary`, `employee_count`, and `average_age`. Keep `department` as a normal column.


````python
employees.groupby("department", as_index=False).agg(
    average_salary=("salary", "mean"),
    highest_salary=("salary", "max"),
    employee_count=("name", "count"),
    average_age=("age", "mean")
)
````

### Exercise 26 — ★★☆
Filter active employees, group them by department, and produce `active_employee_count` and `average_active_salary`.


````python
employees[employees["active"] == True].groupby(
    "department",
    as_index=False
).agg(
    active_employee_count=("name", "count"),
    average_active_salary=("salary", "mean")
)
````
