```python
import duckdb

duckdb.sql("CREATE VIEW duolingo_flagship AS SELECT * FROM read_csv_auto('../../data/duolingo_flagship_v2.csv')")

df = duckdb.sql("SELECT * FROM duolingo_flagship").df()
```

```python
df
```

```python
import pandas as pd
df["practice_time"] = pd.to_datetime(df["practice_time"], errors = "coerce")
```

```python
df["grammar_tags"].astype("category")
```

```python
df["ui_language"] = df["ui_language"].astype("category")
```

```python
df["learning_language"] = df["learning_language"].astype("category")
```

```python
df["pos"] = df["pos"].astype("category")
```

```python
df[df["history_correct"] > df["history_seen"]]
```

```python
df["lemma"] = df["lemma"].str.strip()
```

```python
df["surface_form"] = df["surface_form"].str.strip()
```
